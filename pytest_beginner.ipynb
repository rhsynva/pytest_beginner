```python
%%writefile test_sysexit.py
import pytest
def f():
  raise(SystemExit(1))

def test_mytest():
  with pytest.raises(SystemExit):
    f()
```

```python
!pytest test_sysexit.py
```

```python
%%writefile test_class.py
class TestClass:
  def test_one(self):
    x="this"
    assert "h" in x
  def test_two(self):
    x="hello"
    assert hasattr(x,"check")
```

```python
!pytest test_class.py
```

```python
%%writefile test_class_demo.py
class TestClassDemoInstance:
  value=0
  def test_one(self):
    self.value=1
    assert self.value==1
  def test_two(self):
    assert self.value==1
```

```python
!pytest test_class_demo.py
```

```python
%%writefile test_approx.py
import pytest

def test_sum():
    assert (0.1 + 0.2) == pytest.approx(0.3)
```
